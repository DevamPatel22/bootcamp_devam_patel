# Environment & Config Check

This notebook checks that the Homework 02 environment is set up correctly. It verifies that Python can load environment variables, import NumPy, and find the expected project folders.

```python
from pathlib import Path
import sys
import os

print("Python version:", sys.version)
print("Current working directory:", Path.cwd())
```

```python
PROJECT_ROOT = Path.cwd().parent

print("Project root:", PROJECT_ROOT)
print("Project root exists:", PROJECT_ROOT.exists())
print("Files/folders in project root:")
print([path.name for path in PROJECT_ROOT.iterdir()])
```

```python
from dotenv import load_dotenv

env_path = PROJECT_ROOT / ".env"

print(".env path:", env_path)
print(".env exists:", env_path.exists())

load_dotenv(env_path)

api_key = os.getenv("API_KEY")
data_dir = os.getenv("DATA_DIR")

print("API_KEY present:", api_key is not None)
print("DATA_DIR:", data_dir)
```

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print("Array:", arr)
print("Mean:", arr.mean())
print("Vectorized * 10:", arr * 10)
```
