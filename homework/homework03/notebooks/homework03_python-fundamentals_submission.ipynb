# Homework 03 - Python Fundamentals

This notebook uses NumPy and pandas to explore the Home Credit loan application dataset. The goal is to practice array operations, dataset loading, summary statistics, groupby aggregation, saving outputs, and writing reusable functions.

```python
from pathlib import Path
import time

import numpy as np
import pandas as pd
```

## 1. NumPy Operations

This section creates a NumPy array, performs elementwise operations, and compares a regular Python loop with a vectorized NumPy operation.

```python
numbers = np.array([10, 20, 30, 40, 50])

print("Original array:", numbers)
print("Array plus 5:", numbers + 5)
print("Array multiplied by 2:", numbers * 2)
print("Array squared:", numbers ** 2)
print("Mean:", numbers.mean())
```

```python
large_numbers = np.arange(1_000_000)

start_time = time.time()
loop_result = []
for value in large_numbers:
    loop_result.append(value * 2)
loop_time = time.time() - start_time

start_time = time.time()
vectorized_result = large_numbers * 2
vectorized_time = time.time() - start_time

print("Loop time:", loop_time)
print("Vectorized time:", vectorized_time)
print("First five vectorized results:", vectorized_result[:5])
```

## 2. Dataset Loading

This section loads the Home Credit application training data from the raw data folder and inspects the first few rows.

```python
HOMEWORK_ROOT = Path.cwd().parent
DATA_RAW = HOMEWORK_ROOT / "data" / "raw"
DATA_PROCESSED = HOMEWORK_ROOT / "data" / "processed"

csv_path = DATA_RAW / "application_train.csv"

print("Homework root:", HOMEWORK_ROOT)
print("CSV path:", csv_path)
print("CSV exists:", csv_path.exists())
```

```python
df = pd.read_csv(csv_path)

print("Shape:", df.shape)
df.head()
```

```python
df.info()
```

## 3. Summary Statistics

This section calculates summary statistics for numeric columns and uses groupby aggregation to compare values across categories.

```python
summary_stats = df.describe()

summary_stats
```

```python
contract_summary = df.groupby("NAME_CONTRACT_TYPE")["AMT_CREDIT"].agg(["count", "mean", "median", "min", "max"])

contract_summary
```

```python
summary_path = DATA_PROCESSED / "summary.csv"

summary_stats.to_csv(summary_path)

print("Saved summary statistics to:", summary_path)
```

## 5. Reusable Function

This section defines a reusable function for calculating summary statistics from a DataFrame.

```python
def get_summary_stats(dataframe):
    """Return summary statistics for numeric columns in a DataFrame."""
    return dataframe.describe()
```

```python
get_summary_stats(df)
```
