# Homework 06 - Data Preprocessing

This notebook loads a raw dataset, applies reusable cleaning functions, compares the original and cleaned data, documents assumptions, and saves the cleaned dataset to `data/processed/`.

```python
from pathlib import Path
import sys

import pandas as pd
```

```python
HOMEWORK_ROOT = Path.cwd().parent
DATA_RAW = HOMEWORK_ROOT / "data" / "raw"
DATA_PROCESSED = HOMEWORK_ROOT / "data" / "processed"
SRC_DIR = HOMEWORK_ROOT / "src"

if str(SRC_DIR) not in sys.path:
    sys.path.append(str(SRC_DIR))

from cleaning import fill_missing_median, drop_missing, normalize_data

print("Homework root:", HOMEWORK_ROOT)
print("Raw data folder exists:", DATA_RAW.exists())
print("Processed data folder exists:", DATA_PROCESSED.exists())
```

## 1. Load Raw Dataset

This section loads the raw sample dataset from `data/raw/`.

```python
raw_path = DATA_RAW / "loan_sample.csv"

df_raw = pd.read_csv(raw_path)

print("Shape:", df_raw.shape)
df_raw
```

## 2. Create Missing Values for Cleaning Practice

The sample dataset does not contain missing values, so this section creates a copy with a few missing values to demonstrate the preprocessing functions.

```python
df_working = df_raw.copy()

df_working.loc[1, "income"] = pd.NA
df_working.loc[3, "loan_amount"] = pd.NA

print("Missing values before cleaning:")
print(df_working.isna().sum())

df_working
```

## 3. Apply Cleaning Functions

This section fills missing numeric values with medians, drops any remaining missing rows, and normalizes numeric columns.

```python
numeric_columns = ["income", "loan_amount"]

df_cleaned = fill_missing_median(df_working, numeric_columns)
df_cleaned = drop_missing(df_cleaned)
df_cleaned = normalize_data(df_cleaned, numeric_columns)

print("Missing values after cleaning:")
print(df_cleaned.isna().sum())

df_cleaned
```

## 4. Compare Original and Cleaned Data

This section compares missing values, shape, and numeric summaries before and after cleaning.

```python
print("Original shape:", df_working.shape)
print("Cleaned shape:", df_cleaned.shape)

print("\nMissing values before cleaning:")
print(df_working.isna().sum())

print("\nMissing values after cleaning:")
print(df_cleaned.isna().sum())

print("\nOriginal numeric summary:")
display(df_working[numeric_columns].describe())

print("\nCleaned numeric summary:")
display(df_cleaned[numeric_columns].describe())
```

## 5. Assumptions

- Missing numeric values are filled with the median because the median is less affected by extreme values.
- Any remaining rows with missing values are dropped after median filling.
- Numeric columns are normalized with min-max scaling so they use a common 0 to 1 range.
- This sample dataset is small, so the goal is to demonstrate the preprocessing workflow rather than build a final model-ready dataset.

```python
processed_path = DATA_PROCESSED / "loan_sample_cleaned.csv"

df_cleaned.to_csv(processed_path, index=False)

print("Saved cleaned dataset to:", processed_path)
```
